# Static: add the real hydrogen map (optional)

Out of the box, the Hydrogen station shows the whole sky at 1420 MHz from the GSM2016 model. This notebook swaps in the
real thing: the **HI4PI** survey's map of neutral hydrogen across the entire sky (HI4PI Collaboration 2016, A&A 594, A116).

It takes about 10 minutes, almost all of it downloading the 318 MB survey file. Run the cells in order
(Runtime > Run all works too). You'll be asked to upload two files from your project:

- `tools/make_sky.py`
- `data/manifest.json`

At the end, two files download: `hi4pi.webp` and an updated `manifest.json`.

```python
# Upload tools/make_sky.py and data/manifest.json (select both in the file picker)
import os, shutil
from google.colab import files

os.makedirs("data", exist_ok=True)
uploaded = files.upload()
for name in uploaded:
    if name.endswith(".json"):
        shutil.move(name, "data/manifest.json")
    elif name.endswith(".py"):
        shutil.move(name, "make_sky.py")
assert os.path.exists("make_sky.py") and os.path.exists("data/manifest.json"), "Upload both files, then run this cell again."
print("Ready.")
```

## Put the files in your project

1. Move `hi4pi.webp` into your project's `data/special/` folder.
2. Replace `data/manifest.json` with the downloaded one. (If your browser saved it as `manifest (1).json`, rename it.)
3. Reload the site and tune to 1420.4058 MHz. The hydrogen survey fades in when you lock on, and the About panel credits it.
